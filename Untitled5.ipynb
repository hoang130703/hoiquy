```python
import numpy as np
from sklearn.linear_model import LinearRegression
# Instead of using load_boston, we will fetch the data directly
# as suggested in the error message
import pandas as pd

# Fetch the dataset from the original source
data_url = "http://lib.stat.cmu.edu/datasets/boston"
raw_df = pd.read_csv(data_url, sep="\s+", skiprows=22, header=None)
data = np.hstack([raw_df.values[::2, :], raw_df.values[1::2, :2]])
target = raw_df.values[1::2, 2]

# Now we have the data and target, similar to what load_boston provided
X_boston, y_boston = data, target

# Split the dataset into training and testing sets
train_size = X_boston.shape[0] // 2  # Integer division for the split
X_train = X_boston[:train_size]
X_test = X_boston[train_size:]
y_train = y_boston[:train_size]
y_test = y_boston[train_size:]

# Print the training and testing set sizes
print('Training and testing set sizes:', X_train.shape, X_test.shape)

# Initialize the Linear Regression model
regr = LinearRegression()

# Fit the model to the training data
regr.fit(X_train, y_train)

# Print the coefficients and intercept
print('Coefficients and intercept:')
print('Coefficients:', regr.coef_)
print('Intercept:', regr.intercept_)

# Print the testing score (R-squared)
print('Testing Score:', regr.score(X_test, y_test))

# Calculate and print the Training Mean Squared Error (MSE)
train_mse = np.mean((regr.predict(X_train) - y_train) ** 2)
print('Training MSE:', train_mse)

# Calculate and print the Testing Mean Squared Error (MSE)
test_mse = np.mean((regr.predict(X_test) - y_test) ** 2)
print('Testing MSE:', test_mse)
```
